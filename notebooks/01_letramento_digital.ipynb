## 2. Segurança na internet

**Senha forte** tem pelo menos 12 caracteres, mistura letras, números e símbolos e **não** usa dados pessoais (nome, data de nascimento).

**Phishing** é quando alguém se passa por uma empresa ou pessoa conhecida para roubar seus dados. Desconfie de:
- mensagens com **urgência** ("sua conta será bloqueada hoje!");
- **links estranhos** ou encurtados;
- pedidos de **senha, código do SMS ou Pix**.

Vamos usar o próprio Python para testar se uma senha é forte 👇 (não digite sua senha verdadeira!)

```python
senha = "futebol123"   # troque por um exemplo inventado

tem_tamanho = len(senha) >= 12
tem_numero = any(c.isdigit() for c in senha)
tem_maiuscula = any(c.isupper() for c in senha)
tem_simbolo = any(not c.isalnum() for c in senha)

pontos = tem_tamanho + tem_numero + tem_maiuscula + tem_simbolo
print("Pontuação da senha:", pontos, "de 4")
if pontos == 4:
    print("Senha forte! 💪")
else:
    print("Senha fraca. Tente deixá-la maior e misturar MAIÚSCULAS, números e símbolos.")
```
