## 1. Comparações

| Operador | Significado |
|---|---|
| `==` | igual |
| `!=` | diferente |
| `>` / `<` | maior / menor |
| `>=` / `<=` | maior ou igual / menor ou igual |
| `and`, `or`, `not` | e, ou, não |

```python
nota = 6.5
print(nota >= 6)
print(nota == 10)
print(nota >= 6 and nota < 7)
```

## 2. Condicionais: `if`, `elif`, `else`
Atenção à **indentação** (os 4 espaços no começo da linha): é ela que diz o que está "dentro" do `if`.

```python
media = 5.5

if media >= 7:
    situacao = "Aprovado(a)"
elif media >= 5:
    situacao = "Recuperação"
else:
    situacao = "Reprovado(a)"

print(f"Média {media}: {situacao}")
```

## 3. Listas

```python
notas = [8.0, 6.5, 9.5, 4.0, 7.5]
print("Primeira nota:", notas[0])
print("Quantidade de notas:", len(notas))
print("Maior nota:", max(notas))
print("Média:", sum(notas) / len(notas))
```

## 4. Laço `for`: repetir para cada item

```python
for n in notas:
    if n >= 7:
        print(n, "→ acima da média")
    else:
        print(n, "→ precisa melhorar")
```

```python
# Tabuada do 7 usando range()
for i in range(1, 11):
    print(f"7 x {i} = {7 * i}")
```

## 5. Laço `while`: repetir enquanto uma condição for verdadeira

```python
dinheiro = 0
semana = 0
while dinheiro < 100:
    semana += 1
    dinheiro += 15     # guardando R$ 15 por semana
print(f"Em {semana} semanas você junta R$ {dinheiro}.")
```

## 🎮 Desafio: jogo de adivinhação
O computador sorteia um número de 1 a 20 e você tenta adivinhar.

```python
import random

secreto = random.randint(1, 20)
tentativas = 0
chute = 0

while chute != secreto:
    chute = int(input("Seu chute (1 a 20): "))
    tentativas += 1
    if chute < secreto:
        print("É maior!")
    elif chute > secreto:
        print("É menor!")

print(f"Acertou em {tentativas} tentativa(s)! 🎉")
```
