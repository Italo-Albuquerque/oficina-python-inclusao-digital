## 1. Variáveis

Uma variável é como uma **caixa com etiqueta**: o nome é a etiqueta e o valor é o que está dentro.

```python
nome = "Ana"
idade = 16
altura = 1.62
estuda = True

print(nome, idade, altura, estuda)
```

## 2. Tipos de dados

| Tipo | Significado | Exemplo |
|---|---|---|
| `int` | número inteiro | `16` |
| `float` | número com casas decimais | `1.62` |
| `str` | texto (*string*) | `"Ana"` |
| `bool` | verdadeiro ou falso | `True` |

A função `type()` mostra o tipo de uma variável:

```python
print(type(nome))
print(type(idade))
print(type(altura))
print(type(estuda))
```

## 3. Trabalhando com textos

```python
escola = "Escola Estadual de Fortaleza"
print(escola.upper())      # tudo em maiúsculas
print(escola.lower())      # tudo em minúsculas
print(len(escola))         # quantidade de caracteres
print(escola.split())      # separa as palavras
```

## 4. f-strings: juntando texto e variáveis
Coloque um `f` antes das aspas e as variáveis entre `{ }`.

```python
print(f"{nome} tem {idade} anos e {altura} m de altura.")
print(f"Daqui a 10 anos, {nome} terá {idade + 10} anos.")
```

## 5. Entrada de dados com `input()`

O `input()` **sempre devolve texto**. Para fazer contas, converta com `int()` ou `float()`.

```python
seu_nome = input("Qual é o seu nome? ")
ano_nascimento = int(input("Em que ano você nasceu? "))
print(f"Olá, {seu_nome}! Em 2026 você completa {2026 - ano_nascimento} anos.")
```
