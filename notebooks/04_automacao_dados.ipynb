## 1. Criando um arquivo CSV
CSV é uma planilha em formato de texto: cada linha é um registro e as colunas são separadas por vírgula.
O comando mágico `%%writefile` do Colab salva o conteúdo da célula em um arquivo.

```python
%%writefile notas_turma.csv
aluno,matematica,portugues,ciencias,faltas
Ana,8.5,7.0,9.0,2
Bruno,5.0,6.5,4.5,8
Carla,9.5,9.0,8.5,0
Diego,6.0,5.5,7.0,5
Eduarda,7.5,8.0,6.5,1
Felipe,4.0,5.0,5.5,12
Gabriela,8.0,9.5,9.0,3
Heitor,6.5,7.0,7.5,4
Isabela,10.0,8.5,9.5,0
João,5.5,4.0,6.0,9
```

## 2. Lendo a planilha com pandas

```python
import pandas as pd

df = pd.read_csv("notas_turma.csv")
df
```

```python
print("Formato (linhas, colunas):", df.shape)
df.describe()
```

## 3. Calculando a média de cada aluno

```python
disciplinas = ["matematica", "portugues", "ciencias"]
df["media"] = df[disciplinas].mean(axis=1).round(1)
df
```

## 4. Automatizando a situação de cada aluno
Regra (exemplo): média ≥ 7 e até 10 faltas → **Aprovado**; média entre 5 e 7 → **Recuperação**; caso contrário → **Reprovado**.

```python
def situacao(linha):
    if linha["faltas"] > 10:
        return "Reprovado por falta"
    if linha["media"] >= 7:
        return "Aprovado"
    if linha["media"] >= 5:
        return "Recuperação"
    return "Reprovado"

df["situacao"] = df.apply(situacao, axis=1)
df.sort_values("media", ascending=False)
```

```python
# Quantos alunos em cada situação?
df["situacao"].value_counts()
```

## 5. Filtrando dados

```python
recuperacao = df[df["situacao"] == "Recuperação"]
recuperacao[["aluno", "media"]]
```

## 6. Gráfico das médias

```python
import matplotlib.pyplot as plt

ordenado = df.sort_values("media")
cores = ["#2E7D32" if m >= 7 else "#F9A825" if m >= 5 else "#C62828" for m in ordenado["media"]]

plt.figure(figsize=(8, 4))
plt.barh(ordenado["aluno"], ordenado["media"], color=cores)
plt.axvline(7, linestyle="--", color="gray", label="Média para aprovação")
plt.xlabel("Média")
plt.title("Média final por aluno (dados fictícios)")
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Salvando o boletim automaticamente

```python
df.to_csv("boletim_turma.csv", index=False)
print("Boletim salvo! Abra a pasta 📁 no menu lateral do Colab para baixar o arquivo.")
```
